```sage
# A dictionary to store lists og graphs, keyed by their spectrum
graphs_by_spec = {}
graphs_by_spec_S = {}
n=7
```

```sage
## Iterate through all non-ismorphic graphs on n vertices
# The Graphs() genrator is built-in tool for creating alll non-iso grphs on n vertices
# print(n)

for G in graphs(n) :
    # The spectrum can have floting point inaccuracies
    # To use it as a dictnory key, we round he eigenvalues 
    # and store them in tuple, which is a hashtable object.
    spectrum = tuple(sorted([round(val, 6) for val in G.spectrum ()]))
    # print(G)
    # Add the graph to our dictionary
    if spectrum in graphs_by_spec :
        graphs_by_spec[spectrum].append(G)
        
    else :
        graphs_by_spec[spectrum] = [G]
        # print(graphs_by_spec[spectrum]) 
    
```

```sage
# Now, search for spectra that correspond to more than one graph
GS = graphs_by_spec
print(f"Searching for cospectral mates among graphs with {n} vertices...")
found_mates = False
for spectrum, graph_list in GS.items():
    if len(graph_list) > 1:
        found_mates = True
        print("\n-------------------------------------------")
        print("Found a set of non-isomorphic cospectral graphs!")
        print("Spectrum:", spectrum)
        
        # Display the graphs in the set
        for G in graph_list:
            G.show()
            
        # We can formally verify they are not isomorphic
        G_1 = graph_list[0]
        G_2 = graph_list[1]
        print("Are they isomorphic?", G_1.is_isomorphic(G_2))
        print("-------------------------------------------")

if not found_mates:
    print(f"\nNo cospectral mates found for n={n}. All graphs on {n} vertices are DS.")
```
