```python
def create_G_n_k(n,k):
    Kn = graphs.CompleteGraph(n)
    g = Kn
    for i in range(k) :
        g.add_vertex(n+i)
        g.add_edge(i,n+i)
    return g

def get_spectrum(G):
    """
    Calculates the spectrum (eigenvalues of the adjacency matrix) of a graph.
    The eigenvalues are sorted for consistent comparison.
    
    Args:
        G (sage.graphs.graph.Graph): A SageMath graph object.

    Returns:
        np.ndarray: A sorted array of the graph's eigenvalues.
    """
    # Sagemath's G.spectrum() method is highly optimized and returns
    # a sorted list of floating-point eigenvalues.
    eigenvalues = G.spectrum()
    
    # Convert to numpy array for consistent comparison with np.allclose
    return list(np.array(eigenvalues))

```

```python
# check interlacing for graph G and subgraph H spectrums
def interlacing(G, H):
    m = H.order()
    n = G.order()
    spec_G = get_spectrum(G)
    spec_H = get_spectrum(H)
    #print(f"Spectrum of G: {spec_G}")
    #print(f"Spectrum of H: {spec_H}")
    result = True
    for i in range(m):
        # check interlacing
        # print(f"i={i}, n={len(spec_G)}, m={m}")
        if spec_H[i] <= spec_G[i] and spec_H[i] >= spec_G[n-m+i]:
            #print(f"Interlacing holds for i={i}")
            pass
        else:
            #print(f"Interlacing does not hold for i={i}")
            result = False
            break
    return result
```
