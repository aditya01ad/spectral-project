```python
def create_G_n_k(n,k):
    Kn = graphs.CompleteGraph(n)
    g = Kn
    for i in range(k) :
        g.add_vertex(n+i)
        g.add_edge(i,n+i)
    return g
```

```python
def ch_co(G, filename) :
  n = G.order()
  m = G.size()
  spec_G = [round(x.real(),5) for x in sorted(G.spectrum())]

  with open(filename) as fh:
    for line in fh:
      line = line.strip()
      if not line:
        continue
      H = Graph(line, format='graph6')
      if H.order() != n or H.size() != m :
        continue
      spec_H = [round(x.real(),5) for x in sorted(H.spectrum())]
      if spec_H == spec_G :
        if not G.is_isomorphic(H) :
          print("Found cospectral mate!")
          return False
  return True
```

```python
g = create_G_n_k(6,3)
g.show()
ch_co(g, "graph9.g6")
```
