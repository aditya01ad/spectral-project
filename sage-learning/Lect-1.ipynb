```python
G = C7

# Get the number of vertices and edges
print("Order (vertices) : ", G.order())
print("Size (edges) : ", G.size())
# Get the list of vertices and edges
print("Vertices : ", G.vertices())
print("Edges : ", G.edges(labels=False))
# Get degree info
print("Dgree sequnce : ", G.degree_sequence())
print("Is the Graph regularr?", G.is_regular())
```
